# 03 - Customer Segmentation using K-Means
This notebook segments customers into different value and risk buckets using unsupervised clustering (K-Means).

```python
import os
import sys
import pandas as pd

sys.path.append('../src')
from analysis import run_kmeans_segmentation
```

### Load Enriched Dataset and Run K-Means Clustering

```python
df = pd.read_csv('../data/processed/telco_churn_cleaned.csv')
# First, we need basic feature engineering to compute ChurnRiskScore and NumAddOns
from feature_engineering import run_feature_engineering
df_fe = run_feature_engineering(df)

# Run segmentation
df_seg = run_kmeans_segmentation(df_fe, n_clusters=4)
df_seg['BusinessSegment'].value_counts()
```

### Profiling Customer Segments

```python
profile = df_seg.groupby('BusinessSegment').agg({
    'tenure': 'mean',
    'MonthlyCharges': 'mean',
    'NumAddOns': 'mean',
    'ChurnRiskScore': 'mean',
    'Churn': 'mean'
}).round(2)
profile
```
