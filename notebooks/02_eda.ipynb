# 02 - Exploratory Data Analysis (EDA)
This notebook runs the exploratory data analysis visualizations, answering key business questions about demographics, tenure, charges, and services.

```python
import os
import sys
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.append('../src')
from analysis import compute_kpis
```

### Load Cleaned Data & Compute Top-level Business KPIs

```python
df = pd.read_csv('../data/processed/telco_churn_cleaned.csv')
kpis = compute_kpis(df)
```

### Visualizing Churn by Contract Type

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Contract', hue='Churn', palette='muted')
plt.title('Customer Churn by Contract Type')
plt.ylabel('Number of Customers')
plt.show()
```

### Visualizing Churn by Tenure

```python
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='tenure', hue='Churn', multiple='stack', palette='muted', bins=30)
plt.title('Tenure Distribution by Churn Status')
plt.xlabel('Tenure (months)')
plt.show()
```
