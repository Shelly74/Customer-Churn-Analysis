# 01 - Data Cleaning & Feature Creation Pipeline
This notebook demonstrates the end-to-end data cleaning pipeline used to preprocess the IBM Telco Customer Churn dataset.

```python
import os
import sys
import pandas as pd

# Add src directory to system path
sys.path.append('../src')

from data_cleaning import load_raw_data, fix_data_types, remove_duplicates, standardize_categoricals, investigate_outliers, validate_numerics, create_features
```

### Load Raw Dataset

```python
raw_path = '../data/raw/telco_churn_raw.csv'
df_raw = load_raw_data(raw_path)
df_raw.head()
```

### Run Cleaning Pipeline Steps

```python
df_clean = fix_data_types(df_raw)
df_clean = remove_duplicates(df_clean)
df_clean = standardize_categoricals(df_clean)
investigate_outliers(df_clean)
df_clean = validate_numerics(df_clean)
df_clean = create_features(df_clean)

print("\nCleaned Dataset Shape:", df_clean.shape)
df_clean.head()
```
